## Question 1: Frequency, Relative Frequency, Percentage, and Cumulative Frequency

```python
ratings = ['A', 'B', 'C', 'B', 'A', 'A', 'C', 'B', 'B', 'A',
           'C', 'A', 'B', 'A', 'C', 'B', 'A', 'B', 'A', 'C']

a_count = 0
b_count = 0
c_count = 0

for rating in ratings:
    if rating == 'A':
        a_count = a_count + 1
    elif rating == 'B':
        b_count = b_count + 1
    elif rating == 'C':
        c_count = c_count + 1

total = len(ratings)

a_relative = a_count / total
b_relative = b_count / total
c_relative = c_count / total

print('Rating A')
print('Frequency:', a_count)
print('Relative Frequency:', a_relative)
print('Percentage:', a_relative * 100, '%')
print('Cumulative Frequency:', a_count)
print()

print('Rating B')
print('Frequency:', b_count)
print('Relative Frequency:', b_relative)
print('Percentage:', b_relative * 100, '%')
print('Cumulative Frequency:', a_count + b_count)
print()

print('Rating C')
print('Frequency:', c_count)
print('Relative Frequency:', c_relative)
print('Percentage:', c_relative * 100, '%')
print('Cumulative Frequency:', a_count + b_count + c_count)
```

## Question 2: Discretize a Numeric Variable Into Five Levels

```python
numbers = [
    11.49, 9.59, 12.94, 16.57, 8.30,
    8.30, 16.74, 13.30, 6.31, 10.83,
    10.47, 10.64, 13.24, 7.09, 7.28,
    13.44, 9.99, 8.09, 9.07, 6.30
]

sorted_numbers = sorted(numbers)

print('Sorted Data:')
print(sorted_numbers)
print()

group_size = len(sorted_numbers) // 5

for level in range(5):
    start = level * group_size
    end = start + group_size
    group = sorted_numbers[start:end]
    print('Level', level + 1, ':', group)
```

## Question 3: Standardize the Values Using Z-Scores

```python
import math

values = [
    46.93, 138.07, 64.77, 51.46, 15.38,
    15.37, 11.32, 223.21, 120.23, 19.85,
    19.80, 12.39, 55.79, 86.31, 3.99,
    101.51, 30.71, 94.28, 102.39, 18.04
]

total = 0

for value in values:
    total = total + value

mean = total / len(values)

sum_difference = 0

for value in values:
    difference = value - mean
    sum_difference = sum_difference + (difference * difference)

variance = sum_difference / len(values)
standard_deviation = math.sqrt(variance)

print('Mean:', round(mean, 2))
print('Standard Deviation:', round(standard_deviation, 2))
print()
print('Standardized Values:')

for value in values:
    z_score = (value - mean) / standard_deviation
    print(value, '->', round(z_score, 3))
```
