## Task 1 - Create the Inventory Dictionary

The dictionary stores each product as a key and the available stock as its value.


```python
inventory = {
    "Widget": 10,
    "Gadget": 5,
    "Sensor": 0,
    "Cable": 15
}

print("Starting Inventory:")
print(inventory)
```

## Task 1 - Create the Order Queue

Each order is stored as a list containing the item name and requested quantity.

These sample orders demonstrate:
- full fulfillment,
- partial fulfillment,
- out-of-stock handling,
- and an invalid item.


```python
orders = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 5],
    ["Adapter", 4]
]

print("Incoming Orders:")
for order in orders:
    print(order)
```

## Task 2 - Process the Orders

The program checks every order against the inventory dictionary.

Rules:
1. If stock is enough, fulfill the whole order.
2. If some stock exists but not enough, fulfill the available amount and record the remainder.
3. If stock is zero or the item does not exist, record the entire requested quantity as unfulfilled.


```python
unfulfilled_orders = []
fully_fulfilled_count = 0

print("ORDER PROCESSING")
print("----------------")

for order in orders:
    item = order[0]
    requested_quantity = order[1]

    available_stock = inventory.get(item)

    if available_stock is None:
        print(f"{item}: Item does not exist in inventory. Could not fulfill {requested_quantity}.")
        unfulfilled_orders.append([item, requested_quantity])

    elif available_stock >= requested_quantity:
        inventory[item] = available_stock - requested_quantity
        fully_fulfilled_count = fully_fulfilled_count + 1
        print(f"{item}: Full order fulfilled. Supplied {requested_quantity}.")

    elif available_stock > 0:
        unfulfilled_quantity = requested_quantity - available_stock
        print(
            f"{item}: Partial order fulfilled. "
            f"Supplied {available_stock}, unable to supply {unfulfilled_quantity}."
        )
        inventory[item] = 0
        unfulfilled_orders.append([item, unfulfilled_quantity])

    else:
        print(f"{item}: Out of stock. Could not fulfill {requested_quantity}.")
        unfulfilled_orders.append([item, requested_quantity])
```

## Task 3 - Final Inventory

This section prints the updated stock levels after every order has been processed.


```python
print("FINAL UPDATED INVENTORY")
print("-----------------------")

for item, quantity in inventory.items():
    print(f"{item}: {quantity}")
```

## Task 3 - Summary Report

The assignment requires:
1. Final updated inventory
2. Total number of fully fulfilled orders
3. Items and quantities that could not be fully supplied


```python
print("END-OF-BATCH SUMMARY")
print("--------------------")

print("Final inventory:", inventory)
print("Total fully fulfilled orders:", fully_fulfilled_count)
print("Unfulfilled order items:", unfulfilled_orders)
```

## Complete Source Code

The following cell contains the entire program in one place.  
This is the version you can copy into a `.py` file for your GitHub repository.


```python
inventory = {
    "Widget": 10,
    "Gadget": 5,
    "Sensor": 0,
    "Cable": 15
}

orders = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 5],
    ["Adapter", 4]
]

unfulfilled_orders = []
fully_fulfilled_count = 0

print("STARTING INVENTORY")
print("------------------")
print(inventory)

print()
print("ORDER PROCESSING")
print("----------------")

for order in orders:
    item = order[0]
    requested_quantity = order[1]

    available_stock = inventory.get(item)

    if available_stock is None:
        print(f"{item}: Item does not exist in inventory. Could not fulfill {requested_quantity}.")
        unfulfilled_orders.append([item, requested_quantity])

    elif available_stock >= requested_quantity:
        inventory[item] = available_stock - requested_quantity
        fully_fulfilled_count = fully_fulfilled_count + 1
        print(f"{item}: Full order fulfilled. Supplied {requested_quantity}.")

    elif available_stock > 0:
        unfulfilled_quantity = requested_quantity - available_stock
        print(
            f"{item}: Partial order fulfilled. "
            f"Supplied {available_stock}, unable to supply {unfulfilled_quantity}."
        )
        inventory[item] = 0
        unfulfilled_orders.append([item, unfulfilled_quantity])

    else:
        print(f"{item}: Out of stock. Could not fulfill {requested_quantity}.")
        unfulfilled_orders.append([item, requested_quantity])

print()
print("END-OF-BATCH SUMMARY")
print("--------------------")

print("Final updated inventory:")
for item, quantity in inventory.items():
    print(f"{item}: {quantity}")

print("Total fully fulfilled orders:", fully_fulfilled_count)
print("Unfulfilled order items:", unfulfilled_orders)
```
